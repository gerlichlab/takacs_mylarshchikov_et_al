## Figure 1A

### Set parameters

```python
fig1_region = ('chr2', 120_800_000, 122_800_000)

# hi-c
resolution = 10_000
max_separation = 750_000
hic_vmin = 5e-4
hic_vmax = 1e-1

prometa_path = '/groups/gerlich/labinfo/Papers/2025_ReplChromConf/analysis/assets/coolers_downsampled/Prometa.mcool'
g1_path = '/groups/gerlich/labinfo/Papers/2025_ReplChromConf/analysis/assets/coolers_downsampled/WT_G1.mcool'
g2_path = '/groups/gerlich/labinfo/Papers/2025_ReplChromConf/analysis/assets/coolers_downsampled/WT_G2.mcool'
dsor_path = '/groups/gerlich/labinfo/Papers/2025_ReplChromConf/analysis/assets/coolers_downsampled/dSororin_G2.mcool'
```

### Plot Hi-C

```python
fig1_plotter = RegionPlotter(*fig1_region)

fig1_plotter.add_hic(prometa_path, resolution, hic_vmin, hic_vmax, max_separation, germany)
fig1_plotter.add_hic(g1_path, resolution, hic_vmin, hic_vmax, max_separation, germany)
fig1_plotter.add_hic(g2_path, resolution, hic_vmin, hic_vmax, max_separation, germany)
fig1_plotter.add_hic(dsor_path, resolution, hic_vmin, hic_vmax, max_separation, germany)

fig, axes = fig1_plotter.plot(figsize=(8, 6), dpi=150, tick_all=False, tick_step=100_000)

for i in range(4):
    axes[i,0].set_yticks([]) # change yticks on the Hi-C track
    axes[i,0].set_yticklabels([]) # rename yticks on the Hi-C track
    axes[i,0].xaxis.set_visible(False)  # disable x axis
    axes[i,0].set_xticklabels([])
fig.tight_layout()
```

## Figure 1D

### Set parameters

```python
# chip-seq
fig1_chip_region = ('chr2', 121_000_000, 122_000_000)

binsize = 1000
chip_vmin = None
chip_vmax = None
rad21_g1 = '/groups/gerlich/labinfo/Papers/2025_ReplChromConf/analysis/assets/bams_downsampled/RAD21/WT_G1/RAD21_WT_G1.chip.bw'
rad21_g2 = '/groups/gerlich/labinfo/Papers/2025_ReplChromConf/analysis/assets/bams_downsampled/RAD21/WT_G2/RAD21_WT_G2.chip.bw'
sor_g2 = '/groups/gerlich/experiments/Experiments_006500/006575/bam_repo/Sororin/WT_G2/Sororin_WT_G2.chip.bw'
ctcf_g2 = '/groups/gerlich/experiments/Experiments_006500/006575/bam_repo/CTCF/WT_G2/CTCF_WT_G2.chip.bw'
```

### Plot ChIP-Seq

```python
fig1_plotter = RegionPlotter(*fig1_chip_region)

fig1_plotter.add_fill(rad21_g1, binsize, -0.1, 1.2, 'grey', 'darkblue', mean_subtract=True)
fig1_plotter.add_fill(rad21_g2, binsize, -0.1, 1.2, 'grey', 'darkblue', mean_subtract=True)
fig1_plotter.add_fill(sor_g2, binsize, -0.1, 1.2, 'grey', 'darkred', mean_subtract=True)
fig1_plotter.add_fill(ctcf_g2, binsize, -0.1, 1.4, 'grey', 'green', mean_subtract=True)

fig, axes = fig1_plotter.plot(figsize=(8, 4), dpi=150, tick_all=False, tick_step=100_000)

for i in range(4):
    axes[i, 0].xaxis.set_visible(False) # Remove x axis
    axes[i, 0].spines['bottom'].set_visible(False) # Remove the horizontal line
    axes[i, 0].tick_params(width=0.5, labelsize=6)
    for spine in axes[i, 0].spines.values():
        spine.set_linewidth(0.5)

# Set ticks on y axes
axes[0, 0].set_yticks([0, 1.2])
axes[1, 0].set_yticks([0, 1.2])
axes[2, 0].set_yticks([0, 1.2])
axes[3, 0].set_yticks([0, 1.4])

fig.tight_layout()
```

## Suppl. Fig 1F

```python
binsize = 1000
chip_vmin = None
chip_vmax = None
rad21_g1_r1 = '/groups/gerlich/labinfo/Papers/2025_ReplChromConf/analysis/assets/bams_downsampled/RAD21/WT_G1/bio_rep1/RAD21_WT_G1_BR1.chip.bw'
rad21_g1_r2 = '/groups/gerlich/labinfo/Papers/2025_ReplChromConf/analysis/assets/bams_downsampled/RAD21/WT_G1/bio_rep2/RAD21_WT_G1_BR2.chip.bw'
rad21_r1 = '/groups/gerlich/labinfo/Papers/2025_ReplChromConf/analysis/assets/bams_downsampled/RAD21/WT_G2/bio_rep1/RAD21_WT_G2_BR1.chip.bw'
rad21_r2 = '/groups/gerlich/labinfo/Papers/2025_ReplChromConf/analysis/assets/bams_downsampled/RAD21/WT_G2/bio_rep2/RAD21_WT_G2_BR2.chip.bw'
sor_r1 = '/groups/gerlich/experiments/Experiments_006500/006575/bam_repo/Sororin/WT_G2/bio_rep1/Sororin_WT_G2_BR1.chip.bw'
sor_r2 = '/groups/gerlich/experiments/Experiments_006500/006575/bam_repo/Sororin/WT_G2/bio_rep2/Sororin_WT_G2_BR2.chip.bw'
ctcf_r1 = '/groups/gerlich/experiments/Experiments_006500/006575/bam_repo/CTCF/WT_G2/bio_rep1/CTCF_WT_G2_BR1.chip.bw'
ctcf_r2 = '/groups/gerlich/experiments/Experiments_006500/006575/bam_repo/CTCF/WT_G2/bio_rep2/CTCF_WT_G2_BR2.chip.bw'
```

### Plot ChIP-Seq

```python
# Set tick parameters on y axes
ticks_rad21 = [0, 1.4]
ticks_sor = [0, 1.4]
ticks_ctcf = [0, 2.4]
```

```python
sfig1_region = ('chr2', 120_800_000, 122_800_000)

sfig1_plotter = RegionPlotter(*sfig1_region)

sfig1_plotter.add_fill(rad21_g1_r1, binsize, -0.1, 1.4, 'grey', 'darkblue', mean_subtract=True)
sfig1_plotter.add_fill(rad21_g1_r2, binsize, -0.1, 1.4, 'grey', 'darkblue', mean_subtract=True)
sfig1_plotter.add_fill(rad21_r1, binsize, -0.1, 1.4, 'grey', 'darkblue', mean_subtract=True)
sfig1_plotter.add_fill(rad21_r2, binsize, -0.1, 1.4, 'grey', 'darkblue', mean_subtract=True)
sfig1_plotter.add_fill(sor_r1, binsize, -0.1, 1.4, 'grey', 'darkred', mean_subtract=True)
sfig1_plotter.add_fill(sor_r2, binsize, -0.1, 1.4, 'grey', 'darkred', mean_subtract=True)
sfig1_plotter.add_fill(ctcf_r1, binsize, -0.1, 2.4, 'grey', 'green', mean_subtract=True)
sfig1_plotter.add_fill(ctcf_r2, binsize, -0.1, 2.4, 'grey', 'green', mean_subtract=True)

fig, axes = sfig1_plotter.plot(figsize=(8, 6), dpi=300, tick_all=False, tick_step=100_000)

for i in range(8):
    axes[i, 0].xaxis.set_visible(False) # Remove x axis
    axes[i, 0].spines['bottom'].set_visible(False) # Remove the horizontal line
    axes[i, 0].tick_params(axis='y', length=4, width=0.5, labelsize=6)
    for spine in axes[i, 0].spines.values():
        spine.set_linewidth(0.5)
    
    # Set different ticks based on axis index
    if 0 <= i <= 3:
        axes[i, 0].set_yticks(ticks_rad21)
    elif 4 <= i <= 5:
        axes[i, 0].set_yticks(ticks_sor)
    elif 6 <= i <= 7:
        axes[i, 0].set_yticks(ticks_ctcf)

fig.tight_layout()
```
