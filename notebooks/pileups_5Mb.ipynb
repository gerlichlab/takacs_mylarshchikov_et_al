# Pileups for 5Mb flanks @ 20kb resolution

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

## Load data

```python
repo = Path('/groups/gerlich/labinfo/Papers/2025_ReplChromConf/analysis/assets/pileups/flank_5Mb_res_20Kb/')
```

```python
obs_pups = {cond: {sis: {level: np.load(repo / f"{cond}_G2.{sis}.{level}.obs.pileup.npy")
                         for level in ('high', 'med', 'low')}
                   for sis in ('cis', 'trans')}
            for cond in ('WT', 'dNIPBL', 'dWAPL', 'dCTCF')}

oe_pups = {cond: {sis: {level: np.load(repo / f"{cond}_G2.{sis}.{level}.obsexp.pileup.npy")
                        for level in ('high', 'med', 'low')}
                  for sis in ('cis', 'trans')}
           for cond in ('WT', 'dNIPBL', 'dWAPL', 'dCTCF')}
```

## Parameters

```python
ticks = [0, 250, 500]
ticklabels = ['-5', '0', '5']
```

## Compare CTCF depletion with Wild type

### Figure 3D - dCTCF/WT trans, high

```python
# Plot only high level boundaries
cond1 = 'WT'
cond2 = 'dCTCF'
sis = 'trans'
level = 'high'

fig, axes = plt.subplots(1, 1, figsize=(6, 6), dpi=100)
pup1 = obs_pups[cond1][sis][level]
pup2 = obs_pups[cond2][sis][level]
ratio_pup = np.log2(pup2 / pup1)
g = plt.imshow(ratio_pup, cmap='RdBu_r', vmin=-2, vmax=+2, interpolation='none')
plt.colorbar(g, label=f"{cond2} / {cond1} log2 ratio", shrink=0.6)
plt.title(f"High-level boundaries")
plt.ylabel('dCTCF / WT trans contacts')
plt.xticks(ticks, ticklabels)
plt.yticks([], [])
plt.tick_params(axis='x', length=10, width=0.5)
for spine in axes.spines.values():
        spine.set_linewidth(0.5)
plt.tight_layout()
plt.show()
```
