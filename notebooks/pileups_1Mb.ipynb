# Pilueps with +/- 1Mb window

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from modules.germany import germany
```

```python
plt.rcParams.update({'pdf.fonttype': 42,
                     'ps.fonttype': 42})
sns.set_context('talk')
```

## Load data

```python
repo = Path('/groups/gerlich/labinfo/Papers/2025_ReplChromConf/analysis/assets/pileups/flank_1Mb_res_10Kb/')
```

```python
pileups = {cond: {sis: {level: np.load(repo / f"{cond}_G2.{sis}.{level}.obsexp.pileup.npy")
                        for level in ('high', 'med', 'low')}
                  for sis in ('cis', 'trans')}
           for cond in ('WT', 'dNIPBL', 'dWAPL', 'dCTCF')}
```

# Obs/Exp Pileups

### Figure 2E - Wild type trans contacts

```python
condition = 'WT'
sis = 'trans'
fig, axes = plt.subplots(1, 3, figsize=(17, 10), dpi=150)
for col_idx, (level, pup) in enumerate(pileups[condition][sis].items()):
    ax = axes[col_idx]
    plt.sca(ax)
    g = plt.imshow(np.log2(pup), cmap='RdBu_r', vmin=-.6, vmax=+.6, interpolation='none')
    # Set ticks and labels
    ax.set_yticks([])
    ax.set_yticklabels([])
    ax.set_xticks([0, 100, 200])
    ax.set_xticklabels(['-1', '0', '+1'])
    # Change tick linewidth
    ax.tick_params(width=0.5)
    # Change border linewidth
    for spine in ax.spines.values():
        spine.set_linewidth(0.5)
    plt.title(f'Boundary level: {level}')
    if col_idx == 0:
        ax.set_ylabel('Trans contacts')

plt.tight_layout()
fig.colorbar(g, ax=axes, shrink=0.4)
fig.text(0.29, 0.2, 'Distance to TAD boundaries [Mb]')

plt.show()
```

### Suppl. Figure 2A - Wild type cis contacts

```python
obs_pups = {cond: {sis: {level: np.load(repo / f"{cond}_G2.{sis}.{level}.obs.pileup.npy")
                         for level in ('high', 'med', 'low')}
                   for sis in ('cis', 'trans')}
            for cond in ('WT', 'dNIPBL', 'dWAPL', 'dCTCF')}
```

```python
condition = 'WT'
sis = 'cis'
vmin, vmax = 1e-4, 1e-1

fig, axes = plt.subplots(1, 3, figsize=(17, 10), dpi=150)
for col_idx, (level, pup) in enumerate(obs_pups[condition][sis].items()):
    ax = axes[col_idx]
    plt.sca(ax)
    g = plt.imshow(pup, norm='log', cmap=germany, vmin=vmin, vmax=vmax, interpolation='none')
    # Set ticks and labels
    ax.set_yticks([])
    ax.set_yticklabels([])
    ax.set_xticks([0, 100, 200])
    ax.set_xticklabels(['-1', '0', '+1'])
    # Change tick linewidth
    ax.tick_params(width=0.5)
    # Change border linewidth
    for spine in ax.spines.values():
        spine.set_linewidth(0.5)
    plt.title(f'Boundary level: {level}')
    if col_idx == 0:
        ax.set_ylabel('Cis contacts')

plt.tight_layout()
fig.colorbar(g, ax=axes, shrink=0.4)
fig.text(0.29, 0.2, 'Distance to TAD boundaries [Mb]')

plt.show()
```
